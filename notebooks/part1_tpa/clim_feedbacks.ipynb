### Energy balance

Recall our slab model of the atmosphere from TPA

In an equilibrium energy in = energy out 

$$\frac{S_0}{4}(1-\alpha_p) = \sigma T_s^4 - \epsilon \sigma T_a^4$$

and we recall that $2 T_a^4 = T_s^4$

thus 

$$\frac{S_0}{4}(1-\alpha_p) = \sigma T_s^4 - \frac{\epsilon}{2} \sigma T_s^4 = (1-\frac{\epsilon}{2}) \sigma T_s^4$$

We define $\epsilon_{e} \equiv 1-\frac{\epsilon}{2}$

And thus 

$$\frac{S_0}{4}(1-\alpha_p) = \epsilon_{e} \sigma T_s^4 $$

$\epsilon_{e}$ is the "effective" emissivity of the Earth's surface. 

For today's climate, water vapor and other greenhouse gases in the atmosphere give it an emissivity $\epsilon$ of 0.78.

This translates into an effective emissivity $\epsilon_e$ of the surface of 1-0.78/2=0.61

### Equilibrium Climate Sensitivity (ECS)

We now define the Equilibrium Climate Sensitivity (denoted ECS or $\Delta T_{2xCO2}$):

*The global mean surface warming necessary to balance the planetary energy budget after a doubling of atmospheric CO$_2$.*

The temperature must increase so that the increase in OLR is exactly equal to the radiative forcing:

$$ OLR_f - OLR_{2xCO2} = \Delta R $$
